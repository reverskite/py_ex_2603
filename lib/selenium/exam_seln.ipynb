```python
# -------------------------------------------------
# 1. Selenium이란?
# -------------------------------------------------
"""
Selenium은 브라우저를 사람이 조작하듯이 자동으로 조작할 수 있는 도구입니다.

- 브라우저 실행
- 페이지 이동
- 버튼 클릭
- 입력창에 글자 입력
- 텍스트 가져오기

이런 작업을 코드로 할 수 있습니다.

특히 자바스크립트로 만들어진 '동적 웹페이지'에서 많이 사용합니다.
"""

# -------------------------------------------------
# 2. Selenium 설치 (처음 한 번만)
# -------------------------------------------------
"""
터미널(cmd)에서 아래 명령어 실행:

pip install selenium
"""

# -------------------------------------------------
# 3. 기본 사용 예제
# -------------------------------------------------

# 브라우저를 실행하고 조작하는 객체
from selenium import webdriver

# 요소를 찾을 때 사용하는 선택 방식 지정 도구
from selenium.webdriver.common.by import By

# 1) 브라우저 실행
driver = webdriver.Chrome()

"""
webdriver.Chrome()은
크롬 브라우저를 실행합니다.
"""

# 2) 웹페이지 접속
driver.get("https://[Log in to view URL]")

"""
get()은 특정 웹페이지로 이동하는 명령어입니다.
"""

# 3) 요소 하나 찾기
quote = driver.find_element(By.CLASS_NAME, "text")

"""
find_element는 '요소 하나'를 찾습니다.

By.CLASS_NAME은
class 이름으로 요소를 찾겠다는 의미입니다.

"text"는 해당 사이트에서 명언 문장이 들어있는 class 이름입니다.
"""

# 4) 텍스트 가져오기
print(quote.text)

"""
.text를 사용하면
해당 요소 안에 있는 텍스트만 가져올 수 있습니다.
"""

# -------------------------------------------------
# 4. 브라우저 종료
# -------------------------------------------------
driver.quit()

"""
quit()은 브라우저를 종료합니다.
반드시 마지막에 실행해주는 것이 좋습니다.
"""
```
